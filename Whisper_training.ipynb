````python
# The virtual environment approach in Colab can be problematic for consistent dependency management.
# We will revert to a global installation strategy after a runtime restart to ensure proper numpy versioning.

# --- setup ---
!apt-get -qq update
!apt-get -qq install -y ffmpeg

# Install all Python dependencies. Crucially, install numpy==1.26.4 and scipy directly
# to ensure compatibility. Other packages will be reinstalled to link against this numpy.
!pip install -q -U tqdm transformers numpy==1.26.4 matplotlib scipy datasets tensorboard soundfile librosa torchcodec

import os, json, pathlib, subprocess, hashlib, math, wave, threading
import numpy as np
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from transformers import WhisperProcessor
from concurrent.futures import ThreadPoolExecutor, as_completed

# ----------------------------
# USER SETTINGS
# ----------------------------
MODEL = "openai/whisper-tiny"
TRANSCRIPT_DIR = "/content/drive/MyDrive/Transcriptions"
CHUNKS_DIR = "/content/drive/MyDrive/Record_chunks"

# Whisper label safety
MAX_LABEL_TOKENS = 420

# >>> Your desired duration shape <<<
MEAN_SEC  = 5.0
STD_SEC   = 0.3
SHORT_MIN = 0.5
SHORT_MAX = 29.0
LONG_SEC  = 30.0
LONG_PROB = 0.12

CONTEXT_PAD = 0.20
TARGET_SR = 16000

# HARD CAP: never allow outputs beyond this
MAX_OUT_SECONDS = 30.0

# One-sample-under cap guarantees measured duration <= 30.0
MAX_OUT_FRAMES = int(MAX_OUT_SECONDS * TARGET_SR)
DUR_CAP_SEC = (MAX_OUT_FRAMES - 1) / float(TARGET_SR)   # e.g. 29.9999375 at 16k

# concurrency
FFMPEG_WORKERS = 4

# output manifest
MANIFEST_PATH = str(pathlib.Path(CHUNKS_DIR) / "pairs_manifest.jsonl")

processor = WhisperProcessor.from_pretrained(MODEL)

# ----------------------------
# MANIFEST
# ----------------------------
def load_manifest_jsonl(path):
    processed_jsons = set()
    if not os.path.exists(path):
        return processed_jsons
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                obj = json.loads(line)
                tj = obj.get("transcript_json")
                if tj:
                    processed_jsons.add(tj)
            except json.JSONDecodeError:
                continue
    return processed_jsons

def append_manifest_jsonl(path, new_pairs):
    if not new_pairs:
        return
    pathlib.Path(path).parent.mkdir(parents=True, exist_ok=True)
    with open(path, "a", encoding="utf-8") as f:
        for obj in new_pairs:
            f.write(json.dumps(obj, ensure_ascii=False) + "\n")

# ----------------------------
# FAST HELPERS
# ----------------------------
_ffprobe_cache = {}
_ffprobe_lock = threading.Lock()

def ffprobe_duration_sec(path: str) -> float:
    with _ffprobe_lock:
        if path in _ffprobe_cache:
            return _ffprobe_cache[path]
    cmd = [
        "ffprobe", "-v", "error",
        "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1",
        path
    ]
    out = subprocess.check_output(cmd).decode("utf-8").strip()
    dur = float(out)
    with _ffprobe_lock:
        _ffprobe_cache[path] = dur
    return dur

def wav_duration_sec(path: str):
    try:
        with wave.open(path, "rb") as wf:
            fr = wf.getframerate()
            nf = wf.getnframes()
        return nf / float(fr) if fr > 0 else None
    except Exception:
        return None

def run_ffmpeg_cut_len(in_path: str, out_path: str, start_s: float, dur_s: float, sr: int = TARGET_SR):
    pathlib.Path(out_path).parent.mkdir(parents=True, exist_ok=True)
    if dur_s <= 0:
        return
    cmd = [
        "ffmpeg", "-y",
        "-ss", f"{start_s:.6f}",
        "-t",  f"{dur_s:.6f}",   # -to and -t are mutually exclusive; -t has priority
        "-i", in_path,
        "-ac", "1",
        "-ar", str(sr),
        out_path
    ]
    subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

def safe_id_from_path(p: str, n=10) -> str:
    return hashlib.sha1(p.encode("utf-8")).hexdigest()[:n]

def chunk_filename(audio_path: str, transcript_json_path: str, chunk_index: int) -> str:
    base = pathlib.Path(audio_path).stem
    jid = safe_id_from_path(transcript_json_path, n=10)
    return f"{base}__{jid}_chunk{chunk_index:04d}.wav"

def is_useless_segment(seg: dict) -> bool:
    txt = (seg.get("text") or "").strip()
    if txt in {"", ".", "…"}:
        return True
    nsp = seg.get("no_speech_prob", None)
    if nsp is not None and float(nsp) >= 0.6:
        return True
    s = seg.get("start", None)
    e = seg.get("end", None)
    if s is None or e is None:
        return True
    if float(e) <= float(s):
        return True
    return False

# deterministic RNG per transcript
def rng_for_transcript(transcript_path: str):
    seed = int(hashlib.sha1(transcript_path.encode("utf-8")).hexdigest()[:8], 16)
    return np.random.default_rng(seed)

def sample_target_out_sec(rng):
    if rng.random() < LONG_PROB:
        return LONG_SEC
    x = rng.normal(MEAN_SEC, STD_SEC)
    # cap sampled targets to <= 30 always
    return float(np.clip(x, SHORT_MIN, min(SHORT_MAX, MAX_OUT_SECONDS)))

# ----------------------------
# CHUNK BUILD (batch tokenize + HARD core cap)
# ----------------------------
MAX_CORE_ALLOWED = MAX_OUT_SECONDS - 2.0 * CONTEXT_PAD  # core must fit inside 30s window

def merge_segments_targeted(segments, transcript_rng):
    n = len(segments)
    seg_tok = [0] * n

    useful_idx, useful_txt = [], []
    for i, seg in enumerate(segments):
        if is_useless_segment(seg):
            continue
        t = (seg.get("text") or "").strip().lower()
        if t:
            useful_idx.append(i)
            useful_txt.append(t)

    if useful_txt:
        enc = processor.tokenizer(useful_txt, add_special_tokens=False)
        lens = [len(ids) for ids in enc["input_ids"]]
        for i, L in zip(useful_idx, lens):
            seg_tok[i] = L

    chunks = []
    i = 0
    while i < n:
        if is_useless_segment(segments[i]):
            i += 1
            continue

        target_out = sample_target_out_sec(transcript_rng)
        target_core = max(0.50, target_out - 2.0 * CONTEXT_PAD)

        # never allow core target above what can fit
        target_core = min(target_core, MAX_CORE_ALLOWED)

        cur_start = None
        cur_end = None
        cur_tok = 0
        texts = []

        while i < n:
            seg = segments[i]
            if is_useless_segment(seg):
                i += 1
                continue

            s = float(seg["start"]); e = float(seg["end"])
            t = (seg.get("text") or "").strip()

            if cur_start is None:
                cur_start = s
                cur_end = e

            stoks = seg_tok[i]

            # token cap
            if texts and (cur_tok + stoks) > MAX_LABEL_TOKENS:
                break

            # HARD core cap: don't let core span exceed what can fit in 30s
            proposed_end = e
            if (proposed_end - cur_start) > MAX_CORE_ALLOWED and texts:
                break

            # add seg
            texts.append(t)
            cur_tok += stoks
            cur_end = e
            i += 1

            # stop if hit target core
            if (cur_end - cur_start) >= target_core:
                break

        if texts:
            chunks.append({
                "start": cur_start,
                "end": cur_end,
                "text": " ".join([x.strip() for x in texts]).strip(),
                "target_out_sec": float(target_out),
            })
        else:
            # edge case: single huge segment > MAX_CORE_ALLOWED.
            # We still move forward to avoid infinite loop; we keep it but it'll be cut to 30s max.
            seg = segments[i]
            s = float(seg["start"]); e = float(seg["end"])
            t = (seg.get("text") or "").strip()
            chunks.append({"start": s, "end": min(e, s + MAX_CORE_ALLOWED), "text": t, "target_out_sec": float(target_out)})
            i += 1

    return chunks

def compute_centered_window(audio_total, core_start, core_end, target_out):
    core_start = float(max(0.0, core_start))
    core_end   = float(min(audio_total, core_end))
    if core_end < core_start:
        core_end = core_start

    core_span = core_end - core_start

    # enforce <= 30 always, even if transcript span is longer
    target_out = float(min(target_out, MAX_OUT_SECONDS))

    # ensure window contains (as much of) core as possible but never exceeds cap
    need = min(MAX_OUT_SECONDS, max(target_out, core_span + 2*CONTEXT_PAD))
    need = min(need, MAX_OUT_SECONDS)

    mid = 0.5 * (core_start + core_end)
    start = mid - need / 2.0
    start = max(0.0, min(start, audio_total - need))
    return start, need

# ----------------------------
# MAIN
# ----------------------------
TRANSCRIPT_DIR_P = pathlib.Path(TRANSCRIPT_DIR)
CHUNKS_DIR_P = pathlib.Path(CHUNKS_DIR)
CHUNKS_DIR_P.mkdir(parents=True, exist_ok=True)

processed_jsons = load_manifest_jsonl(MANIFEST_PATH)
json_files = sorted(TRANSCRIPT_DIR_P.glob("*.json"))
new_json_files = [p for p in json_files if str(p) not in processed_jsons]

print("Found JSONs:", len(json_files))
print("Already processed (manifest):", len(processed_jsons))
print("Processing JSONs now:", len(new_json_files))

tasks = []
new_pairs = []
bad_json = 0
bad_audio = 0

for jf in tqdm(new_json_files, desc="Building chunk tasks"):
    jf_str = str(jf)
    try:
        obj = json.loads(jf.read_text(encoding="utf-8"))
        audio_path = obj["input_file"]["path"]
        segments = obj["groq_response"]["segments"]
    except Exception:
        bad_json += 1
        continue

    if not os.path.exists(audio_path):
        bad_audio += 1
        continue

    rng = rng_for_transcript(jf_str)
    chunks = merge_segments_targeted(segments, rng)
    if not chunks:
        continue

    for i, ch in enumerate(chunks):
        out_name = chunk_filename(audio_path, jf_str, i)
        out_wav = str(CHUNKS_DIR_P / out_name)

        rec = {
            "audio_path": out_wav,
            "raw_transcription": ch["text"],
            "source_audio": audio_path,
            "chunk_index": i,
            "transcript_json": jf_str,
            "duration_sec_target": float(min(ch["target_out_sec"], MAX_OUT_SECONDS)),
            "sr": TARGET_SR,
            "model": MODEL,
        }
        new_pairs.append(rec)

        if not os.path.exists(out_wav):
            tasks.append((audio_path, out_wav, ch["start"], ch["end"], ch["target_out_sec"]))

print(f"\nFFmpeg tasks (missing wavs): {len(tasks)} | workers={FFMPEG_WORKERS}")

cut_ok = 0
cut_fail = 0

def _cut_task(t):
    audio_path, out_wav, core_s, core_e, target_out = t
    total = ffprobe_duration_sec(audio_path)
    start_s, dur_s = compute_centered_window(total, core_s, core_e, target_out)

    # FINAL HARD CAP (one-sample-under) so measured WAV duration can't exceed 30
    dur_s = min(dur_s, DUR_CAP_SEC)
    if start_s + dur_s > total:
        start_s = max(0.0, total - dur_s)

    run_ffmpeg_cut_len(audio_path, out_wav, start_s, dur_s, sr=TARGET_SR)

    # Safety: if still >30 due to weirdness, recut once
    d = wav_duration_sec(out_wav)
    if d is not None and d > MAX_OUT_SECONDS:
        run_ffmpeg_cut_len(audio_path, out_wav, start_s, DUR_CAP_SEC, sr=TARGET_SR)

    return True

if tasks:
    with ThreadPoolExecutor(max_workers=FFMPEG_WORKERS) as ex:
        futs = [ex.submit(_cut_task, t) for t in tasks]
        for f in tqdm(as_completed(futs), total=len(futs), desc="Cutting wavs"):
            try:
                f.result()
                cut_ok += 1
            except Exception:
                cut_fail += 1

append_manifest_jsonl(MANIFEST_PATH, new_pairs)

print("\nDone.")
print("Bad JSON:", bad_json, "| Missing audio:", bad_audio)
print("Pairs recorded:", len(new_pairs))
print("FFmpeg cut ok:", cut_ok, "| fail:", cut_fail)
print("Manifest:", MANIFEST_PATH)

# ----------------------------
# QUICK CHECK: any wav > 30?
# ----------------------------
def iter_wavs(root):
    stack = [root]
    while stack:
        d = stack.pop()
        try:
            with os.scandir(d) as it:
                for e in it:
                    if e.is_dir(follow_symlinks=False):
                        stack.append(e.path)
                    elif e.is_file(follow_symlinks=False) and e.name.lower().endswith(".wav"):
                        yield e.path
        except (FileNotFoundError, PermissionError):
            continue

too_long = []
for p in iter_wavs(CHUNKS_DIR):
    d = wav_duration_sec(p)
    if d is not None and d > MAX_OUT_SECONDS:
        too_long.append((p, d))

print(f"\nWAVs longer than {MAX_OUT_SECONDS}s: {len(too_long)}")
if too_long[:5]:
    print("Examples:")
    for p, d in too_long[:5]:
        print(d, p)

# Deactivating virtual environment if it was created
# (Note: In Colab, `!source deactivate` might not work as expected across cells)
# !deactivate
````

# analysis of distribution of duration of files in the record chunks

````python
import os, math, wave
import numpy as np
import matplotlib.pyplot as plt
from multiprocessing import Pool, cpu_count

# ---- SETTINGS ----
CHUNKS_DIR = "/content/drive/MyDrive/Record_chunks"  # change if needed
MIN_SEC = 2.0
MAX_SEC = 30.0

# Histogram resolution
BIN_WIDTH = 0.25  # seconds per bin (0.25 = detailed; 0.5 = faster/cleaner)
SMOOTH_SIGMA_BINS = 2.0  # smoothing strength in "bins"

# Multiprocessing
WORKERS = max(1, cpu_count() - 1)
CHUNKSIZE = 256  # increase (512/1024) if you have a TON of files

# ---- FAST recursive scandir (much faster than rglob for huge trees) ----
def iter_wavs(root):
    stack = [root]
    while stack:
        d = stack.pop()
        try:
            with os.scandir(d) as it:
                for e in it:
                    if e.is_dir(follow_symlinks=False):
                        stack.append(e.path)
                    elif e.is_file(follow_symlinks=False) and e.name.lower().endswith(".wav"):
                        yield e.path
        except (FileNotFoundError, PermissionError):
            continue

def wav_duration_sec(path):
    # Reads just the header; extremely fast.
    try:
        with wave.open(path, "rb") as wf:
            fr = wf.getframerate()
            nf = wf.getnframes()
        if fr <= 0:
            return None
        return nf / float(fr)
    except Exception:
        return None

def gaussian_smooth(y, sigma_bins):
    if sigma_bins <= 0:
        return y
    radius = int(max(3, math.ceil(3 * sigma_bins)))
    x = np.arange(-radius, radius + 1)
    k = np.exp(-(x**2) / (2 * sigma_bins**2))
    k /= k.sum()
    return np.convolve(y, k, mode="same")

# ---- RUN ----
wav_paths = list(iter_wavs(CHUNKS_DIR))
print(f"Found {len(wav_paths):,} wav files under: {CHUNKS_DIR}")

if not wav_paths:
    raise SystemExit("No .wav files found. Check CHUNKS_DIR.")

with Pool(processes=WORKERS) as pool:
    durs = list(pool.imap_unordered(wav_duration_sec, wav_paths, chunksize=CHUNKSIZE))

durs = np.array([d for d in durs if d is not None], dtype=np.float32)

print(f"Readable durations: {len(durs):,} / {len(wav_paths):,}")
if len(durs) == 0:
    raise SystemExit("Could not read durations from any wav. Files might be corrupt or not WAV PCM.")

# Stats
def pct(x): return float(np.percentile(durs, x))
outside = np.sum((durs < MIN_SEC) | (durs > MAX_SEC))
print(
    f"Duration stats (sec):\n"
    f"  min={durs.min():.3f}  p50={pct(50):.3f}  mean={durs.mean():.3f}  "
    f"p90={pct(90):.3f}  p95={pct(95):.3f}  p99={pct(99):.3f}  max={durs.max():.3f}\n"
    f"  outside [{MIN_SEC}, {MAX_SEC}] sec: {outside:,} ({outside/len(durs)*100:.2f}%)"
)

# Histogram + density curve
hi = max(MAX_SEC, float(durs.max()))
bins = np.arange(0.0, hi + BIN_WIDTH, BIN_WIDTH)

hist, edges = np.histogram(durs, bins=bins, density=True)
centers = 0.5 * (edges[:-1] + edges[1:])
smooth = gaussian_smooth(hist, SMOOTH_SIGMA_BINS)

plt.figure()
plt.hist(durs, bins=bins, density=True, alpha=0.35)
plt.plot(centers, smooth, linewidth=2)
plt.axvline(MIN_SEC, linestyle="--")
plt.axvline(MAX_SEC, linestyle="--")
plt.xlabel("Duration (seconds)")
plt.ylabel("Density")
plt.title(f"Duration distribution (n={len(durs):,}) | bin={BIN_WIDTH}s | workers={WORKERS}")
plt.show()
````

**Reasoning**:
The next step is to define the `compute_partially_encoder` function, which is crucial for handling dynamic audio contexts during training. This function will be extracted directly from the `finetune.ipynb` notebook to ensure consistency with the original project's methodology.



````python
import torch
from torch import nn

# Whisper decoder hard limit is 448 positions; stay under that for safety.
MAX_AUDIO_CONTEXT = 420

def compute_partially_encoder(model, data, n_audio_ctx):
  diffy = 2*n_audio_ctx - data.shape[2]

  if diffy > 0:
    data = nn.functional.pad(data, [0, diffy, 0, 0, 0, 0], "constant", 0.0)
  elif diffy < 0:
    # In the original notebook, this line was `data = data[:,:,:diffy]`.
    # This is incorrect if diffy is negative.
    # It should be `data = data[:, :, :2*n_audio_ctx]` to truncate to the desired audio context.
    data = data[:, :, :2*n_audio_ctx]

  if n_audio_ctx == MAX_AUDIO_CONTEXT:
    return model.encoder(data).last_hidden_state

  input_embeds = nn.functional.gelu(model.encoder.conv1(data))
  input_embeds = nn.functional.gelu(model.encoder.conv2(input_embeds))
  input_embeds = input_embeds.permute(0, 2, 1)

  embed_pos = model.encoder.embed_positions.weight[:n_audio_ctx]

  hidden_states = input_embeds + embed_pos
  hidden_states = nn.functional.dropout(hidden_states, p=model.encoder.dropout, training=model.encoder.training)

  for idx, encoder_layer in enumerate(model.encoder.layers):
    to_drop = False
    if model.encoder.training:
      dropout_probability = torch.rand([])
      if dropout_probability < model.encoder.layerdrop:
        to_drop = True

    if to_drop:
        layer_outputs = (None, None)
    else:
        if model.encoder.gradient_checkpointing and model.encoder.training:
            layer_outputs = model.encoder._gradient_checkpointing_func(
                encoder_layer.__call__,
                hidden_states,
                None,
                None,
                False,
            )
        else:
            layer_outputs = encoder_layer(
                hidden_states,
                None,
                layer_head_mask=None,
                output_attentions=False,
            )

        hidden_states = layer_outputs[0]

  hidden_states = model.encoder.layer_norm(hidden_states)
  return hidden_states

print("compute_partially_encoder function defined.")
````

````python
import torch
from torch import nn
# from torch.utils.tensorboard import SummaryWriter # Assuming this is needed later for logging

def compute_hidden_state_loss(model_train, model_base, optimizer, criterion, example):
  optimizer.zero_grad()

  # calculate audio context based on length, similar to finetune.ipynb
  n_ctx = int(round((1500.0 / 30.0) * example["length"] ))

  # Add some randomness to the audio context length
  extra_ctx = torch.randint(-min(64, n_ctx // 3), min(64, n_ctx // 3), (1,)).item()
  n_ctx += extra_ctx

  input_features = example["input_features"].to(device) # Move to device
  input_ids = torch.tensor([example["input_ids"]], dtype=torch.long).to(device) # Move to device

  encoder_hidden_states_partial = compute_partially_encoder(model_train, input_features, n_ctx)
  output_partial = model_train.decoder(
      input_ids=input_ids,
      encoder_hidden_states=encoder_hidden_states_partial,
      output_hidden_states=True
  )

  with torch.no_grad():
    encoder_hidden_states_full = compute_partially_encoder(model_base, input_features, 1500)
    output_full = model_base.decoder(
        input_ids=input_ids,
        encoder_hidden_states=encoder_hidden_states_full,
        output_hidden_states=True
    )

  # Calculate loss using concatenated hidden states as per finetune.ipynb
  loss = criterion(
      torch.cat(output_partial.hidden_states, 0),
      torch.cat(output_full.hidden_states, 0)
    )

  loss.backward()
  optimizer.step()

  return loss

print("compute_hidden_state_loss function defined.")
````

The training loop previously failed due to a tensor dimension mismatch. I will now redefine the `compute_hidden_state_loss` function to cap `n_ctx` to the maximum allowed encoder context length and ensure it's at least 1, as per the instructions, which should prevent such errors.



````python
import torch
from torch import nn

def compute_hidden_state_loss(model_train, model_base, optimizer, criterion, example):
  optimizer.zero_grad()

  # calculate audio context based on length, similar to finetune.ipynb
  n_ctx = int(round((1500.0 / 30.0) * example["length"] ))

  # Add some randomness to the audio context length
  extra_ctx = torch.randint(-min(64, n_ctx // 3), min(64, n_ctx // 3), (1,)).item()
  n_ctx += extra_ctx

  # Ensure n_ctx is within valid bounds for embed_positions.weight
  # model_train.encoder.embed_positions.weight.shape[0] gives the maximum possible context length
  max_embed_positions_length = model_train.encoder.embed_positions.weight.shape[0]
  n_ctx = min(n_ctx, max_embed_positions_length)
  n_ctx = max(1, n_ctx) # Ensure n_ctx is at least 1

  input_features = example["input_features"].to(device) # Move to device
  input_ids = torch.tensor([example["input_ids"]], dtype=torch.long).to(device) # Move to device

  encoder_hidden_states_partial = compute_partially_encoder(model_train, input_features, n_ctx)
  output_partial = model_train.decoder(
      input_ids=input_ids,
      encoder_hidden_states=encoder_hidden_states_partial,
      output_hidden_states=True
  )

  with torch.no_grad():
    encoder_hidden_states_full = compute_partially_encoder(model_base, input_features, 1500)
    output_full = model_base.decoder(
        input_ids=input_ids,
        encoder_hidden_states=encoder_hidden_states_full,
        output_hidden_states=True
    )

  # Calculate loss using concatenated hidden states as per finetune.ipynb
  loss = criterion(
      torch.cat(output_partial.hidden_states, 0),
      torch.cat(output_full.hidden_states, 0)
    )

  loss.backward()
  optimizer.step()

  return loss

print("compute_hidden_state_loss function redefined with corrections.")
````

**Reasoning**:
The `compute_hidden_state_loss` function has been redefined. Now, as per the instructions, I will execute the main training loop, which involves iterating through the `final_dataset`, processing each example, calculating loss, and logging metrics.



````python
from tqdm.auto import tqdm

num_length = 0
step = 0

for epoch in range(8):
  pbar = tqdm(final_dataset.shuffle(seed=epoch), desc=f"Epoch {epoch}")
  for example in pbar:
    example = get_sample(example)
    if example["length"] > 29.0: continue

    loss = compute_hidden_state_loss(model_train, model_base, optimizer, criterion, example)
    step += 1
    num_length += example["length"]

    # writer.add_scalar("loss/train", loss.item(), step)
    # writer.add_scalar("length/train", num_length, step)
    # writer.add_scalar("epoch/train", epoch, step)

    pbar.set_description(f"Epoch {epoch}, Loss: {loss.item():.4f}")

print("Training loop execution complete.")
````
