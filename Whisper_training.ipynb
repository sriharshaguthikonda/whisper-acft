split the audio files into ~30 second files with ~5 second overlapping segments into another folder and maintain the mother file name. it has to be non destructive to the mother audio files! plan in such a way so we are able to the map the text to each of the split audio files correctly!


````python
# --- setup ---
!apt-get -qq update
!apt-get -qq install -y ffmpeg
!pip -q install -U tqdm transformers datasets[audio] torchcodec

import os, json, pathlib, subprocess
from tqdm.auto import tqdm
from transformers import WhisperProcessor

# Change these:
MODEL = "openai/whisper-tiny"  # or your chosen Whisper checkpoint
TRANSCRIPT_DIR = "/content/drive/My Drive/Transcriptions"   # where your .json live
CHUNKS_DIR = "/content/drive/My Drive/Record_chunks"  # output folder for chunk WAVs

processor = WhisperProcessor.from_pretrained(MODEL)

# Whisper decoder hard limit is 448 positions; stay under that for safety.
MAX_LABEL_TOKENS = 420
MAX_CHUNK_SECONDS = 25.0     # keep comfortably under 30s
PAD_SECONDS = 0.15           # pad to avoid chopping phonemes

def run_ffmpeg_cut(in_path, out_path, start_s, end_s, sr=16000):
    pathlib.Path(out_path).parent.mkdir(parents=True, exist_ok=True)
    cmd = [
        "ffmpeg", "-y",
        "-ss", str(start_s),
        "-to", str(end_s),
        "-i", in_path,
        "-ac", "1",
        "-ar", str(sr),
        out_path
    ]
    subprocess.run(cmd, check=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

def is_useless_segment(seg):
    txt = (seg.get("text") or "").strip()
    if txt in {"", ".", "…"}:
        return True
    # if present, skip likely-silence segments (your file shows this field exists) :contentReference[oaicite:3]{index=3}
    nsp = seg.get("no_speech_prob", None)
    if nsp is not None and float(nsp) >= 0.6:
        return True
    return False

def merge_segments_into_chunks(segments):
    chunks = []
    cur = []
    cur_start = None
    cur_end = None

    def flush():
        nonlocal cur, cur_start, cur_end
        if not cur:
            return
        text = " ".join([s["text"].strip() for s in cur]).strip()
        if text:
            chunks.append({"start": cur_start, "end": cur_end, "text": text})
        cur, cur_start, cur_end = [], None, None

    for seg in segments:
        if is_useless_segment(seg):
            continue

        s = float(seg["start"])
        e = float(seg["end"])
        t = seg.get("text", "").strip()

        if cur_start is None:
            cur_start = s
        cur_end = e
        cur.append({"start": s, "end": e, "text": t})

        dur = cur_end - cur_start
        tok_len = len(processor.tokenizer.encode(" ".join([x["text"] for x in cur]).lower()))

        if dur >= MAX_CHUNK_SECONDS or tok_len >= MAX_LABEL_TOKENS:
            flush()

    flush()
    return chunks

data_pairs = []

json_files = sorted(pathlib.Path(TRANSCRIPT_DIR).glob("*.json"))
print("Found JSONs:", len(json_files))

for jf in tqdm(json_files, desc="Building chunks from transcripts"):
    obj = json.loads(jf.read_text(encoding="utf-8"))

    # Your JSON has this structure :contentReference[oaicite:4]{index=4}
    audio_path = obj["input_file"]["path"]

    # And segments live here :contentReference[oaicite:5]{index=5}
    segments = obj["groq_response"]["segments"]

    chunks = merge_segments_into_chunks(segments)

    base = pathlib.Path(audio_path).stem
    for i, ch in enumerate(chunks):
        start = max(0.0, ch["start"] - PAD_SECONDS)
        end   = ch["end"] + PAD_SECONDS

        out_wav = str(pathlib.Path(CHUNKS_DIR) / f"{base}_chunk{i:04d}.wav")
        run_ffmpeg_cut(audio_path, out_wav, start, end, sr=16000)

        data_pairs.append({
            "audio_path": out_wav,
            "raw_transcription": ch["text"],
            "source_audio": audio_path,
            "chunk_start": start,
            "chunk_end": end,
            "chunk_index": i,
            "transcript_json": str(jf),
        })

print("Total chunks created:", len(data_pairs))
print("Example:", data_pairs[0])
````

## Create Custom Dataset Loader using datasets[audio]

### Subtask:
Transform the `processed_dataset` into a format fully compatible with the `get_sample` function in `finetune.ipynb`.


**Reasoning**:
I need to define a function `extract_audio_features` that will process each example in the `processed_dataset` to properly extract the audio array and sampling rate, convert the audio to mono, and reformat it into a dictionary with 'array' and 'sampling_rate' keys, along with the 'raw_transcription'. Then, I will apply this function to the `processed_dataset` using `.map()` to create the `final_dataset`.



````python
import numpy as np

def extract_audio_features(example):
    audio_decoder = example['audio']

    # Get AudioSamples object
    samples = audio_decoder.get_all_samples()

    # Extract waveform (torch.Tensor) and sampling rate
    wave = samples.data
    sr = samples.sample_rate

    # Convert to 1D numpy array (mono)
    # .mean(0) averages across channels if stereo, otherwise it's already mono
    wave_mono = wave.mean(0).cpu().numpy()

    # Construct the new audio dictionary
    audio_dict = {
        'array': wave_mono,
        'sampling_rate': sr
    }

    return {
        'audio': audio_dict,
        'raw_transcription': example['raw_transcription']
    }

# Apply the function to processed_dataset to create final_dataset
# The batched=False is important here since get_all_samples() expects one example at a time
final_dataset = processed_dataset.map(extract_audio_features)

print(final_dataset)
print("\nFirst element of final_dataset:")
print(final_dataset[0])
print("\nStructure of 'audio' key in the first element:")
print(final_dataset[0]['audio'])

````

## Define get_sample Function

### Subtask:
Define the `get_sample` function. This function will take an example from the dataset, extract the audio waveform and sampling rate, process it using the `WhisperProcessor`, and encode the raw transcription into token IDs, preparing the data for the model.


**Reasoning**:
To define the `get_sample` function as instructed, I will create a code block that encapsulates all the specified steps for processing an audio example, including extracting waveform, sampling rate, generating input features, encoding transcription, and calculating audio length.



````python
def get_sample(example):
    waveform = example["audio"]["array"]
    sampling_rate = example["audio"]["sampling_rate"]

    # Use the model and processor to transcribe the audio:
    input_features = processor(
        waveform, sampling_rate=sampling_rate, return_tensors="pt"
    ).input_features

    input_ids = processor.tokenizer.encode(example["raw_transcription"].lower())

    return {
        "length": len(waveform) / sampling_rate,
        "input_features": input_features,
        "input_ids": input_ids
    }

print("get_sample function defined.")
````

````python
import torch
from torch import nn

# Whisper decoder hard limit is 448 positions; stay under that for safety.
MAX_AUDIO_CONTEXT = 420

def compute_partially_encoder(model, data, n_audio_ctx):
  diffy = 2*n_audio_ctx - data.shape[2]

  if diffy > 0:
    data = nn.functional.pad(data, [0, diffy, 0, 0, 0, 0], "constant", 0.0)
  elif diffy < 0:
    # In the original notebook, this line was `data = data[:,:,:diffy]`.
    # This is incorrect if diffy is negative.
    # It should be `data = data[:, :, :2*n_audio_ctx]` to truncate to the desired audio context.
    data = data[:, :, :2*n_audio_ctx]

  if n_audio_ctx == MAX_AUDIO_CONTEXT:
    return model.encoder(data).last_hidden_state

  input_embeds = nn.functional.gelu(model.encoder.conv1(data))
  input_embeds = nn.functional.gelu(model.encoder.conv2(input_embeds))
  input_embeds = input_embeds.permute(0, 2, 1)

  embed_pos = model.encoder.embed_positions.weight[:n_audio_ctx]

  hidden_states = input_embeds + embed_pos
  hidden_states = nn.functional.dropout(hidden_states, p=model.encoder.dropout, training=model.encoder.training)

  for idx, encoder_layer in enumerate(model.encoder.layers):
    to_drop = False
    if model.encoder.training:
      dropout_probability = torch.rand([])
      if dropout_probability < model.encoder.layerdrop:
        to_drop = True

    if to_drop:
        layer_outputs = (None, None)
    else:
        if model.encoder.gradient_checkpointing and model.encoder.training:
            layer_outputs = model.encoder._gradient_checkpointing_func(
                encoder_layer.__call__,
                hidden_states,
                None,
                None,
                False,
            )
        else:
            layer_outputs = encoder_layer(
                hidden_states,
                None,
                layer_head_mask=None,
                output_attentions=False,
            )

        hidden_states = layer_outputs[0]

  hidden_states = model.encoder.layer_norm(hidden_states)
  return hidden_states

print("compute_partially_encoder function defined.")
````
