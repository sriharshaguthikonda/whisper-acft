# Clean up Record_chunks directory
This notebook deletes (if present) and recreates the `Record_chunks` folder in Google Drive.


```python
import os
import shutil

# Define the chunks directory in Drive
CHUNKS_DIR = '/content/drive/MyDrive/Record_chunks'

print(f'Attempting to delete and recreate: {CHUNKS_DIR}')

# Delete if it exists
if os.path.exists(CHUNKS_DIR):
    try:
        shutil.rmtree(CHUNKS_DIR)
        print(f'Successfully deleted existing directory: {CHUNKS_DIR}')
    except OSError as e:
        print(f'Error: {e.filename} - {e.strerror}.')
else:
    print(f'Directory does not exist, no need to delete: {CHUNKS_DIR}')

# Recreate empty directory
try:
    os.makedirs(CHUNKS_DIR)
    print(f'Successfully recreated empty directory: {CHUNKS_DIR}')
except OSError as e:
    print(f'Error creating directory {CHUNKS_DIR}: {e}')
```
